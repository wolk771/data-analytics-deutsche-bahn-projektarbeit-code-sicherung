# Dokumentenerstellung aus Jupyter-Notebooks

## Zweck dieses Notebooks

Dieses Notebook dient der zentralen Erstellung von Dokumenten aus den
Projekt-Notebooks.

Dabei werden insbesondere:

- HTML-Dateien für die Integration in andere Notebooks,
- PDF-Dateien für Abgabe, Dokumentation oder Präsentation und
- bei Bedarf temporäre LaTeX-Dateien

automatisch erzeugt und in den zentral definierten Projektordnern gespeichert.

Die Dateipfade werden nicht fest im Notebook eingetragen, sondern aus
`project_config.py` übernommen. Dadurch können andere Notebooks auf dieselben
Dateien zugreifen, unabhängig davon, aus welchem Notebook sie erstellt wurden.

Außerdem werden nach Bedarf bestimmte Stellen in den Projekt-Notebooks mit 
`remove_cell`, `remove_output` oder `remove_input` **Cell Tags** versehen.

> Die Quelldateien der Notebooks werden durch den Export nicht verändert.

> Hinweis: Soll dieses Notebook auf einem anderen Rechner ausgeführt
> werden, müssen alle notwendige Applikationen (wie z.B. MiKTeX) auf dem PC vorhanden sein
>  sowie Zusammenstellungen für Jupiter Lab App bereits eingerichtet werden.

```python
from pathlib import Path
import subprocess
import sys

from project_config import (
    PROJECT_ROOT,
    HTML_DIR,
    PDF_DIR,
    PANDAS_ANALYSIS_HTML,
    PANDAS_ANALYSIS_PDF,
    FRAMEWORK_ANALYSIS_HTML,
    FRAMEWORK_ANALYSIS_PDF,
    PYSPARK_ANALYSIS_HTML,
    PYSPARK_ANALYSIS_PDF,
    STADA_API_HTML,
    STADA_API_PDF,
    TIMETABLE_API_HTML,
    TIMETABLE_API_PDF,
    MAIN_PROJECT_HTML,
    MAIN_PROJECT_PDF
)
```

## Notebook `01_Dateiformate_Pandas.ipynb`

```python
NOTEBOOK_01 = PROJECT_ROOT / "01_Dateiformate_Pandas.ipynb"

print("Quelldatei:", NOTEBOOK_01)
print("HTML-Ziel:", PANDAS_ANALYSIS_HTML)
print("PDF-Ziel:", PANDAS_ANALYSIS_PDF)
```

```python
assert NOTEBOOK_01.exists(), (
    f"Notebook nicht gefunden: {NOTEBOOK_01}"
)

PANDAS_ANALYSIS_HTML.parent.mkdir(
    parents=True,
    exist_ok=True,
)

command = [
    sys.executable,
    "-m",
    "jupyter",
    "nbconvert",
    "--to",
    "html",
    str(NOTEBOOK_01),
    "--output-dir",
    str(PANDAS_ANALYSIS_HTML.parent),
    "--output",
    PANDAS_ANALYSIS_HTML.stem,

    # Tag-Entfernung nur für diesen HTML-Export
    "--TagRemovePreprocessor.enabled=True",
    "--TagRemovePreprocessor.remove_cell_tags=remove_cell",
    "--TagRemovePreprocessor.remove_input_tags=remove_input",
    "--TagRemovePreprocessor.remove_all_outputs_tags=remove_output",
]

result = subprocess.run(
    command,
    cwd=PROJECT_ROOT,
    capture_output=True,
    text=True,
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("HTML-Erstellung fehlgeschlagen.")

print(f"HTML erfolgreich erstellt:\n{PANDAS_ANALYSIS_HTML}")
```

```python
if PANDAS_ANALYSIS_HTML.exists():
    size_kb = PANDAS_ANALYSIS_HTML.stat().st_size / 1024

    print(f"Datei vorhanden: {PANDAS_ANALYSIS_HTML}")
    print(f"Dateigröße: {size_kb:.1f} KB")
else:
    raise FileNotFoundError(
        f"HTML wurde nicht erstellt: {PANDAS_ANALYSIS_HTML}"
    )
```

```python
PANDAS_ANALYSIS_PDF.parent.mkdir(
    parents=True,
    exist_ok=True,
)

cmd = [
    sys.executable,
    "-m",
    "jupyter",
    "nbconvert",
    "--to",
    "pdf",
    str(NOTEBOOK_01),
    "--output-dir",
    str(PANDAS_ANALYSIS_PDF.parent),
    "--output",
    PANDAS_ANALYSIS_PDF.stem,
]

result = subprocess.run(
    cmd,
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)

print(result.stdout)
print(result.stderr)

if result.returncode != 0:
    raise RuntimeError(
        "PDF-Export fehlgeschlagen."
    )
```

```python
if not PANDAS_ANALYSIS_PDF.exists():
    raise FileNotFoundError(
        f"PDF wurde nicht gefunden: {PANDAS_ANALYSIS_PDF}"
    )

print(f"PDF erfolgreich erstellt: {PANDAS_ANALYSIS_PDF}")
print(f"Dateigröße: {PANDAS_ANALYSIS_PDF.stat().st_size / 1024**2:.2f} MB")
```
